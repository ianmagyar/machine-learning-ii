## Q-učenie

Q-učenie (*Q-Learning*), sa učí priamo z interakcií s prostredím, jeho priebeh teda závisí od priebehu konkrétnych epizód. Q-Learning zároveň nevyžaduje známy model prostredia, prechody medzi stavmi sa buď nasimulujú, alebo určia zo skutočnej interakcie agenta s prostredím. Algoritmus zvyčajne pracuje s $\varepsilon$-greedy politikou, kde sa s najväčšou pravdepodobnosťou vyberie akcia, ktorú agent vníma ako najlepšiu možnú, pritom ale existuje možnosť výberu náhodnej akcie. Q-learning je ale off-policy metóda: to znamená, že pri aktualizácii odhadu hodnôt dvojice stav-akcia sa neberie do úvahy práve platná politika agenta.
